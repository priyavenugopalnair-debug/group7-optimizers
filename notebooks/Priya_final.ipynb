```python
import numpy as np
import pandas as pd

# Helper function to generate and save synthetic datasets if they don't exist
W_STAR = np.array([3., 2.])

def make(k, n=200, seed=42):
    r = np.random.default_rng(seed); A = r.standard_normal((n, 2)); A -= A.mean(0)
    Q, _ = np.linalg.qr(A); X = np.sqrt(n) * Q @ np.diag([1, 1/np.sqrt(k)])
    return X, X @ W_STAR

files = {1: "round_k1.csv", 10: "moderate_k10.csv", 100: "narrow_k100.csv"}

# Pre-generate and save files to prevent FileNotFoundError
for kappa, name in files.items():
    X, y = make(kappa)
    df_temp = pd.DataFrame(X, columns=["x1", "x2"])
    df_temp["y"] = y
    df_temp.to_csv(name, index=False)

w0 = np.zeros(2)                      # start at (0, 0)

for kappa, name in files.items():
    d = pd.read_csv(name)
    X = d[["x1", "x2"]].values        # features (ignore sample_id)
    y = d["y"].values
    n = len(X)

    f0 = 0.5 * np.mean((X @ w0 - y) ** 2)      # starting loss
    g0 = X.T @ (X @ w0 - y) / n                # starting gradient
    ratio = abs(g0[0]) / abs(g0[1])            # w1 pull vs w2 pull

    print(f"kappa={kappa:>3}  f(0)={f0:.3f}  g0=({g0[0]:.3f}, {g0[1]:.3f})  ratio={ratio:.1f}x")
```

```python
import numpy as np, matplotlib.pyplot as plt
W_STAR = np.array([3., 2.])

def make(k, n=200, seed=42):
    r = np.random.default_rng(seed); A = r.standard_normal((n, 2)); A -= A.mean(0)
    Q, _ = np.linalg.qr(A); X = np.sqrt(n) * Q @ np.diag([1, 1/np.sqrt(k)])
    return X, X @ W_STAR

def gd_path(X, y, tol=1e-6):
    n = len(y); lam = np.linalg.eigvalsh(X.T @ X / n); a = 2 / (lam[0] + lam[-1])
    w = np.zeros(2); path = [w.copy()]
    while np.linalg.norm(X.T @ (X @ w - y) / n) >= tol:
        w = w - a * X.T @ (X @ w - y) / n; path.append(w.copy())
    return np.array(path)

fig, axes = plt.subplots(1, 3, figsize=(15, 4.5))
for ax, k in zip(axes, (1, 10, 100)):
    X, y = make(k); P = gd_path(X, y)
    g1, g2 = np.meshgrid(np.linspace(-1, 5, 300), np.linspace(-1, 5, 300))
    ax.contour(g1, g2, 0.5*((g1-3)**2 + (g2-2)**2/k), levels=12, colors="#5B6275")
    ax.plot(P[:, 0], P[:, 1], "o-", ms=3, color="#E04F5F", label=f"GD ({len(P)-1} steps)")
    ax.plot(0, 0, "ks", label="start (0,0)"); ax.plot(3, 2, "g*", ms=14, label="minimum (3,2)")
    ax.set_title(f"κ = {k}"); ax.set_xlabel("w1"); ax.set_ylabel("w2"); ax.legend()
plt.tight_layout(); plt.savefig("fig_contours_gd.png", dpi=300)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

W_STAR = np.array([3., 2.])

def make(k, n=200, seed=42):
    r = np.random.default_rng(seed)
    A = r.standard_normal((n, 2))
    A -= A.mean(0)
    Q, _ = np.linalg.qr(A)
    X = np.sqrt(n) * Q @ np.diag([1, 1 / np.sqrt(k)])
    return X, X @ W_STAR

def gd_path(X, y, tol=1e-6):
    n = len(y)
    lam = np.linalg.eigvalsh(X.T @ X / n)
    a = 2 / (lam[0] + lam[-1])
    w = np.zeros(2)
    path = [w.copy()]
    while np.linalg.norm(X.T @ (X @ w - y) / n) >= tol:
        w = w - a * X.T @ (X @ w - y) / n
        path.append(w.copy())
    return np.array(path)

# In Colab the file is uploaded to /content. In the repo it is notebooks/lbfgs_path.csv
lb = pd.read_csv("/content/lbfgs_path.csv")   # columns: iter, w1, w2, kappa

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
titles = {1: "κ = 1: round bowl, as wide as long",
          10: "κ = 10: valley about 3× longer than wide",
          100: "κ = 100: valley 10× longer than wide"}
g1, g2 = np.meshgrid(np.linspace(-1, 7, 400), np.linspace(-1, 7, 400))

for ax, k in zip(axes, (1, 10, 100)):
    X, y = make(k)
    G = gd_path(X, y)
    L = lb[lb.kappa == k].sort_values("iter")[["w1", "w2"]].to_numpy()
    loss = 0.5 * ((g1 - 3) ** 2 + (g2 - 2) ** 2 / k)
    ax.contour(g1, g2, loss, levels=12, colors="#9aa3b5", linewidths=0.8)
    ax.plot(G[:, 0], G[:, 1], "-o", color="#E04F5F", ms=2, lw=0.8,
            label=f"GD: {len(G) - 1} steps")
    ax.plot(L[:, 0], L[:, 1], "-s", color="#1A2237", ms=6, lw=2.5,
            label=f"L-BFGS: {len(L) - 1} steps")
    ax.plot(0, 0, "s", mfc="white", mec="black", ms=8, label="start (0,0)")
    ax.plot(3, 2, "*", color="green", ms=16, label="minimum (3,2)")
    ax.set_xlim(-1, 7)
    ax.set_ylim(-1, 7)
    ax.set_title(titles[k], fontsize=10)
    ax.set_xlabel("w1 (steep direction)")
    ax.set_ylabel("w2 (flat direction)")
    ax.legend(loc="upper right", fontsize=8)

plt.tight_layout()
plt.savefig("fig_contours_gd.png", dpi=200)
plt.show()
```
