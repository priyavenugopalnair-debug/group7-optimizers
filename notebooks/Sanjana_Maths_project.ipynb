This notebook checks the original speaking plan, proposes a **6:30 rehearsal plan with a real 30-second buffer**, and records the team's preliminary optimizer counts for consistency. It does **not** run GD or L-BFGS; Jyoti owns those experiments.

**Shared setup:** three noise-free synthetic regression datasets (κ = 1, 10, 100), 200 rows each, model `y = 3*x1 + 2*x2`, initial weights `(0, 0)`, and stopping rule `L2 gradient norm < 1e-6`. All reported optimizer results remain **preliminary**.

```python
# Speaker allocations, in seconds.
original = {"Priya": 75, "Saranya": 90, "Varuna": 90, "Jyoti": 105, "Sanjana": 60}
rehearsal = {"Priya": 70, "Saranya": 80, "Varuna": 80, "Jyoti": 100, "Sanjana": 60}
limit_seconds = 7 * 60

def fmt(seconds):
    return f"{seconds // 60}:{seconds % 60:02d}"

original_total = sum(original.values())
rehearsal_total = sum(rehearsal.values())
buffer_seconds = limit_seconds - rehearsal_total
print("Original plan:", fmt(original_total))
print("Rehearsal target:", fmt(rehearsal_total))
print("Presentation limit:", fmt(limit_seconds))
print("Actual planned buffer:", buffer_seconds, "seconds")
for name in original:
    print(f"{name:8s}: {fmt(original[name])} → {fmt(rehearsal[name])}")
assert original_total == 420
assert rehearsal_total == 390
assert buffer_seconds == 30
```
