# Sanjana - Group 8: Slide 5 timing calculation

This notebook verifies the five speakers’ planned times and calculates the rehearsal buffer for the November 2 progress presentation. It does **not** run the optimizer experiments; those belong to Jyoti.

All optimizer results on the slide are **preliminary**.

```python
speaker_seconds = {
    "Priya": 75,
    "Saranya": 90,
    "Varuna": 90,
    "Jyoti": 105,
    "Sanjana": 60,
}

total_seconds = sum(speaker_seconds.values())
limit_seconds = 7 * 60
rehearsal_target = 6 * 60 + 30
print(f"Planned total: {total_seconds//60}:{total_seconds%60:02d}")
print(f"Time limit: {limit_seconds//60}:{limit_seconds%60:02d}")
print(f"Target rehearsal: {rehearsal_target//60}:{rehearsal_target%60:02d}")
print(f"Desired buffer: {limit_seconds-rehearsal_target} seconds")
assert total_seconds == limit_seconds
assert limit_seconds-rehearsal_target == 30
```
